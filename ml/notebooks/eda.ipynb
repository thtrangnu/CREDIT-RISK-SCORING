## Section 1 — DAYS_EMPLOYED anomaly

`DAYS_EMPLOYED` is the number of days employed, counted **backwards from the application date** (negative = normal).  
But ~55k rows hold the value **+365243**, roughly 1000 years in the future, which is a hidden encoding for **"retired / not working"**.  
Miss this and every feature derived from the column is completely wrong.


```python
ANOMALY_FLAG = 365243

# --- overall distribution ---
print("=== DAYS_EMPLOYED — top 5 values ===")
print(df["DAYS_EMPLOYED"].value_counts().head())

anomaly_mask = df["DAYS_EMPLOYED"] == ANOMALY_FLAG
n_anomaly = anomaly_mask.sum()
print(f"\nRows anomaly (={ANOMALY_FLAG}): {n_anomaly:,}  "
      f"({n_anomaly/len(df)*100:.1f}% of the dataset)")

# --- default rate: anomaly vs normal ---
dr_anomaly = df.loc[anomaly_mask, "TARGET"].mean()
dr_normal  = df.loc[~anomaly_mask, "TARGET"].mean()
print(f"\nDefault rate — anomaly group : {dr_anomaly:.4f} ({dr_anomaly*100:.1f}%)")
print(f"Default rate — normal  group : {dr_normal:.4f} ({dr_normal*100:.1f}%)")
print("-> the anomaly group defaults more, so this is real signal rather than noise")

# --- DAYS_EMPLOYED distribution after dropping the anomaly ---
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

normal_vals = df.loc[~anomaly_mask, "DAYS_EMPLOYED"] / -365
axes[0].hist(normal_vals, bins=50, edgecolor="none", color="steelblue", alpha=0.8)
axes[0].set_title("Years of employment (normal group)")
axes[0].set_xlabel("Years")
axes[0].set_ylabel("Applications")

# default rate per group: anomaly vs non-anomaly
labels = ["Normal", "Anomaly\n(365243)"]
rates  = [dr_normal, dr_anomaly]
colors = ["steelblue", "tomato"]
axes[1].bar(labels, rates, color=colors, width=0.4)
axes[1].axhline(df["TARGET"].mean(), color="gray", linestyle="--", label="overall rate")
axes[1].set_title("Default rate: anomaly vs normal")
axes[1].set_ylabel("Default rate")
axes[1].legend()
for i, v in enumerate(rates):
    axes[1].text(i, v + 0.002, f"{v:.3f}", ha="center", fontsize=10)

plt.tight_layout()
plt.show()

print("\n-> How Block 2 handles it: set a DAYS_EMPLOYED_ANOM=1 flag, "
      "then set DAYS_EMPLOYED=NaN for that group before computing features.")
```

## Section 2 — Default rate by categorical feature

The three categorical columns with the clearest lift that domain knowledge can explain:
- `NAME_CONTRACT_TYPE`: loan contract type
- `NAME_EDUCATION_TYPE`: education level
- `NAME_INCOME_TYPE`: income source

Goal: see which groups default more than average. This is **univariate lift**, not causation.


```python
overall_rate = df["TARGET"].mean()
cat_focus = ["NAME_CONTRACT_TYPE", "NAME_EDUCATION_TYPE", "NAME_INCOME_TYPE"]

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for ax, col in zip(axes, cat_focus):
    stats = (df.groupby(col, observed=True)["TARGET"]
               .agg(default_rate="mean", count="count")
               .sort_values("default_rate", ascending=True))

    bars = ax.barh(stats.index, stats["default_rate"],
                   color=["tomato" if r > overall_rate else "steelblue"
                          for r in stats["default_rate"]])
    ax.axvline(overall_rate, color="black", linestyle="--", linewidth=1,
               label=f"overall {overall_rate:.3f}")

    # annotate each bar with its count and rate
    for bar, (_, row) in zip(bars, stats.iterrows()):
        ax.text(bar.get_width() + 0.002,
                bar.get_y() + bar.get_height() / 2,
                f"{row['default_rate']:.3f}  (n={row['count']:,})",
                va="center", fontsize=8)

    ax.set_title(col.replace("NAME_", ""), fontsize=10)
    ax.set_xlabel("Default rate")
    ax.legend(fontsize=8)
    ax.set_xlim(0, stats["default_rate"].max() * 1.45)

plt.suptitle("Default rate by group (red = above average)", y=1.02)
plt.tight_layout()
plt.show()

# print the numbers as a table for easier reading
for col in cat_focus:
    print(f"\n--- {col} ---")
    tbl = (df.groupby(col, observed=True)["TARGET"]
             .agg(default_rate="mean", count="count")
             .sort_values("default_rate", ascending=False))
    tbl["lift"] = (tbl["default_rate"] / overall_rate).round(2)
    print(tbl.round(4).to_string())
```

## Section 3 — Top numeric features correlated with TARGET

Pearson correlation with TARGET shows which features carry the strongest **univariate** signal.  
This is not model feature importance, just raw linear signal.

Goal: explain why an application-only baseline already reaches **AUC ~0.74**, and which
signal the multi-table work in Block 2 will add on top.


```python
exclude = {"TARGET", "SK_ID_CURR"}
num_cols_clean = [c for c in df.select_dtypes(include="number").columns
                  if c not in exclude]

corr = (df[num_cols_clean + ["TARGET"]]
        .corr()["TARGET"]
        .drop("TARGET")
        .abs()
        .sort_values(ascending=False))

TOP_N = 15
top = corr.head(TOP_N)

fig, ax = plt.subplots(figsize=(8, 6))
colors = ["tomato" if c.startswith("EXT_SOURCE") else
          "darkorange" if c.startswith("DAYS") else "steelblue"
          for c in top.index]
bars = ax.barh(top.index[::-1], top.values[::-1], color=colors[::-1])
ax.set_xlabel("|Pearson corr| with TARGET")
ax.set_title(f"Top {TOP_N} numeric features by correlation with TARGET")
for bar, val in zip(bars, top.values[::-1]):
    ax.text(bar.get_width() + 0.002, bar.get_y() + bar.get_height() / 2,
            f"{val:.3f}", va="center", fontsize=9)
ax.set_xlim(0, top.max() * 1.2)

# manual legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor="tomato",     label="EXT_SOURCE (credit bureau score)"),
                   Patch(facecolor="darkorange",  label="DAYS_* (time)"),
                   Patch(facecolor="steelblue",   label="Other")]
ax.legend(handles=legend_elements, loc="lower right", fontsize=8)
plt.tight_layout()
plt.show()

print("Top 15 features (|corr| with TARGET):")
print(top.round(4).to_string())
print(f"\n-> EXT_SOURCE_1/2/3 dominate the top; these are external bureau credit scores.")
print(f"-> DAYS_BIRTH (age) is strong too, as seen earlier in the EDA.")
print(f"-> Multi-table work will add signal from borrowing history (bureau, installments, ...)"
      f" that the application table simply does not contain.")
```
